# Synthetic — founder/haplotype-probability kernels

```python
import numpy as np, vigor
rng=np.random.default_rng(42);N,V,F=120,250,8
P=rng.dirichlet(np.ones(F),size=N*V).reshape(N,V,F).astype("float32")
K=vigor.founder_kernel(P)
print("founder probability tensor",P.shape,"kernel",K.shape,"diag mean",np.diag(K).mean())
```
